## **3. Numeric Operation**

```r
x <- 5
y <- 3

sum <- x + y
subtract <- x - y
mult <- x * y
div <- x / y

print(paste0("The + is ", sum))
print(paste0("The - is ", subtract))
print(paste0("The * is ", mult))
print(paste0("The / is ", div))

cat(paste0("\nThe + is ", sum, "\n"))
cat(paste0("The - is ", subtract, "\n"))
cat(paste0("The * is ", mult, "\n"))
cat(paste0("The / is ", div, "\n"))
```

## **8. Vectors**

```r
grades <- c(75, 80, 90, 100)


cat(paste0("Mean: ", mean(grades) ))
cat(paste0("\nMAX: ", max(grades) ))
cat(paste0("\nMIN: ", min(grades) ))
cat(paste0("\nSUM: ", sum(grades) ))
cat(paste0("\nLENGTH: ", length(grades) ))
```
