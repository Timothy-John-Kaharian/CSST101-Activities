## **3. Basic Calculations**

```r
# Student data
quizScore <- 85
laboratory <- 90
project <- 88
exam <- 80

# Grading weight
quizWeight = 0.20
laboratoryWeight <- 0.25
projectWeight <- 0.25
examWeight <- 0.30

weightedQuiz <- quizScore * quizWeight
weightedLab <- laboratory * laboratoryWeight
weightedProject <- project * projectWeight
weightedExam <- exam * examWeight

finalWeightedGrade <- weightedQuiz + weightedLab + weightedProject + weightedExam

cat(sprintf("Student weighted final grade: %.2f%%", finalWeightedGrade))
```

## **5. Using AND and OR**

```r
attendance <- 72
grade <- 70


if (attendance && grade < 75) {
  print("Student is at risk on BOTH attendance and grade below 75")
}

if (attendance || grade < 75){
  print("Student is at risk on EITHER attendance and grade below 75")
}
```

## **12. Add More Rules**

```r
attendance1 <- 65
grade1 <- 71

attendance2 <- 90
grade2 <- 72

attendance3 <- 95
grade3 <- 90

classify_risk <- function(attendance, grade) {

  if (attendance < 75 && grade < 75) {
    print("High Risk")
  } else if (attendance < 85 || grade < 80) {
    print("Moderate Risk")
  } else if (attendance >= 80 && grade >= 80) {
    print("Low Risk")
  }
}


  classify_risk(attendance1, grade1)
  classify_risk(attendance2, grade2)
  classify_risk(attendance3, grade3)
  classify_risk(70,70)
```

## **13. Apply the Rule of Multiple Students**

## **15. Add a Recommendation**

```r
recommendation_risk <- function(risk){
  if (risk == "High Risk"){
    return ("Immediate academic intervention")
  } else if (risk == "Moderate Risk"){
    return ("Academic monitoring")
  } else if (risk == "Low Risk"){
    return ("Continue regular monitoring")
  }
}


recommendation_risk("High Risk")
```

## **17. MINI KRR System**

```r
studentID1 <- "S001"
attendance1 <- 65
grade1 <- 71

studentID2 <- "S002"
attendance2 <- 90
grade2 <- 72

studentID3 <- "S003"
attendance3 <- 95
grade3 <- 90

classify_risk <- function(attendance, grade) {

  if (attendance < 75 && grade < 75) {
    return ("High Risk")
  } else if (attendance < 85 || grade < 80) {
    return ("Moderate Risk")
  } else if (attendance >= 80 && grade >= 80) {
    return ("Low Risk")
  }
}


studentRiskClass1 <- classify_risk(attendance1, grade1)
studentRiskClass2 <-classify_risk(attendance2, grade2)
studentRiskClass3 <-classify_risk(attendance3, grade3)

cat("\n")

recommend_risk <- function(risk){
  if (risk == "High Risk"){
    return ("Immediate academic intervention")
  } else if (risk == "Moderate Risk"){
    return ("Academic monitoring")
  } else if (risk == "Low Risk"){
    return ("Continue regular monitoring")
  }
}


studentRecommendation1 <- recommend_risk(studentRiskClass1)
studentRecommendation2 <- recommend_risk(studentRiskClass2)
studentRecommendation3 <- recommend_risk(studentRiskClass3)



cat(paste0("What is the academic risk of this student, and what intervention should be provided? \n\n"))

cat(paste0("\n-----------------------------------------"))
cat(paste0("\nStudent ID: ", studentID1))
cat(sprintf("\nStudent Grade: %.2f%%", grade1))
cat(sprintf("\nStudent Attendance: %.2f%%", attendance1))
cat(paste0("\n\n", "Student Academic Risk: ", studentRiskClass1))
cat(paste0("\n\n", "Student Recommendation/Intervention: \n", studentRecommendation1))
cat(paste0("\n-----------------------------------------\n"))

cat(paste0("\n-----------------------------------------"))
cat(paste0("\nStudent ID: ", studentID2))
cat(sprintf("\nStudent Grade: %.2f%%", grade2))
cat(sprintf("\nStudent Attendance: %.2f%%", attendance2))
cat(paste0("\n\n", "Student Academic Risk: ", studentRiskClass2))
cat(paste0("\n\n", "Student Recommendation/Intervention: \n", studentRecommendation2))
cat(paste0("\n-----------------------------------------\n"))

cat(paste0("\n-----------------------------------------"))
cat(paste0("\nStudent ID: ", studentID2))
cat(sprintf("\nStudent Grade: %.2f%%", grade3))
cat(sprintf("\nStudent Attendance: %.2f%%", attendance3))
cat(paste0("\n\n", "Student Academic Risk: ", studentRiskClass3))
cat(paste0("\n\n", "Student Recommendation/Intervention: \n", studentRecommendation3))
cat(paste0("\n-----------------------------------------\n"))
```
